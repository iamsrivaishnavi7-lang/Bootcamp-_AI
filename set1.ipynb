```python
a=[56,78,90,65,34,37,26,58,12,57]
print(a)
```

```python
#find max
max=a[0]
for i in range(1,len(a),1):
    if a[i]>max:
        max=a[i]
        id = i
        print(max,i)  
    else:
        continue
print(f"the maximum is {max} at index {id}")



```

```python
#find max withoutindexes
max = a[0]
for j in a:
     if j>max:
         max=j
print(f"the maximum is: {max} ")
```

```python
#find min
min =a[0]
for i in range(1,len(a),1):
    if a[i] < min: 
        min=a[i]
        id = i
        print(min,i)
    else:
          continue
print(f"the minimum is {min} at index {id}")

```
